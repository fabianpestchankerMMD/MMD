```
#datos=pd.read_csv("https://www.wolframcloud.com/obj/d0c0084e-0b60-46db-a4d9-beb33412905e")
#datos.head(10)
datos=pd.read_csv("mushrooms.csv")
datos.head(10)
```

Convierto las etiquetas en numeros

```
encoder=LabelEncoder()
for column in datos.columns:
    datos[column] = encoder.fit_transform(datos[column])
datos.head()
```

```

# se definen los intervalos para los cuartiles
perc =[.25, .50, .75, 1]


# visualizando los datos con describe
desc = datos.describe(percentiles = perc, include='int')

desc
```

Igual que cuando hicimos el ejemplo con R, vemos que hay una columna que tiene solamente un valor. La eliminamos.

```
datos.drop(columns=['veil_type'], axis=1, inplace=True)
datos
```

Ahora hay que crear los juegos de entrenamiento y prueba, hay una función para eso

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(datos.drop(columns=['type']), datos['type'], test_size=0.33, random_state=42)
```

```
X_train.head()
```

Se crea un objeto con el clasificador y con el método **fit** se ajusta el modelo con los datos de entrenamiento

```
clf = DecisionTreeClassifier()
clf = clf.fit(X_train, y_train)
```

A partir del modelo y el método **predict** se crean los valores predichos por el modelo con los datos de prueba, el resultado se asigna a X_pred.

```
y_pred=clf.predict(X_test)
```

```
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(f"Precisión del modelo: {accuracy:.2f}")
```
